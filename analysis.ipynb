```python
import os
import pandas as pd
from PIL import Image

DATA_DIR = "/content/my_dataset"

FASC_IMG_DIR = os.path.join(
    DATA_DIR,
    "fasc_imgs_v1",
    "fasc_images_new_model_v1"
)

FASC_MASK_DIR = os.path.join(
    DATA_DIR,
    "fasc_masks_v1",
    "fasc_masks_new_model_v1"
)

APO_IMG_DIR = os.path.join(
    DATA_DIR,
    "apo_imgs_v1",
    "apo_images_new_model_v1"
)

APO_MASK_DIR = os.path.join(
    DATA_DIR,
    "apo_masks_v1",
    "apo_masks_new_model_v1"
)

TEST_IMG_DIR = os.path.join(
    DATA_DIR,
    "test_images_v2",
    "test_set_v2"
)

print("fasc images:", len(os.listdir(FASC_IMG_DIR)))
print("fasc masks :", len(os.listdir(FASC_MASK_DIR)))
print("apo images :", len(os.listdir(APO_IMG_DIR)))
print("apo masks  :", len(os.listdir(APO_MASK_DIR)))
print("test images:", len(os.listdir(TEST_IMG_DIR)))
```

```python
fasc_imgs = set(os.listdir(FASC_IMG_DIR))
fasc_masks = set(os.listdir(FASC_MASK_DIR))

apo_imgs = set(os.listdir(APO_IMG_DIR))
apo_masks = set(os.listdir(APO_MASK_DIR))

print("fasc image only:")
print(sorted(fasc_imgs - fasc_masks))

print("fasc mask only:")
print(sorted(fasc_masks - fasc_imgs))

print("apo image only:")
print(sorted(apo_imgs - apo_masks))

print("apo mask only:")
print(sorted(apo_masks - apo_imgs))
```
